```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


xls = pd.ExcelFile('Road_Accident_Cleaned_15Rows.xlsx')
df = pd.read_excel('Road_Accident_Cleaned_15Rows.xlsx', sheet_name='Cleaned_Data')


sns.set_theme(style="whitegrid")


city_casualties = df.groupby('City')['Casualties'].sum().reset_index()

plt.figure(figsize=(9, 5))
sns.barplot(data=city_casualties, x='City', y='Casualties', palette='Reds_r')
plt.title('Total Casualties by City')
plt.xlabel('City')
plt.ylabel('Total Casualties')
plt.tight_layout()
plt.show()


plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Severity', palette='Set2')
plt.title('Accident Severity Count')
plt.xlabel('Severity Level')
plt.ylabel('Count')
plt.tight_layout()
plt.show()


plt.figure(figsize=(6, 5))
numeric_df = df[['Latitude', 'Longitude', 'Vehicles_Involved', 'Casualties']]
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
